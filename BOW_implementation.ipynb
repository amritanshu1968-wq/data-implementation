```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
#movie review Dataset
reviews = [
    "I love this movie",
    "this movie is fantastic",
    "I hate this movie",
    "this movie is boring"
]
```

```python
# create BOW model

vectorizer = CountVectorizer(stop_words='english')
```

```python
vectorizer
```

```python
#fit and transform

X = vectorizer.fit_transform(reviews)
```

```python
#view vocabulary

print(vectorizer.get_feature_names_out())
```

```python
#convert to DataFrame
bow = pd.DataFrame(X.toarray(),
                    columns=vectorizer.get_feature_names_out())

print(bow)
```

```python
#predicting a new movie review
new_review = ["I love fantastic movie"]
```

```python
new_vector = vectorizer.transform(new_review)

print(new_vector.toarray())
```

```python
# Import TfidfVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Import pandas for DataFrame
import pandas as pd

# Sample employee reviews
reviews = [
    "Company provides good work environment",
    "Good salary and work culture",
    "Work pressure is high",
    "Excellent work environment"
]

# Create TF-IDF object
vectorizer = TfidfVectorizer()

# Learn vocabulary and create TF-IDF matrix
X = vectorizer.fit_transform(reviews)

# Convert sparse matrix into DataFrame
df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

# Display result
print(df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Documents
documents = [
    "apple banana mango",
    "banana mango",
    "banana orange"
]

# Ignore words appearing in only one document
vectorizer = TfidfVectorizer(min_df=2)

# Create TF-IDF matrix
X = vectorizer.fit_transform(documents)

# Convert into DataFrame
df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print(df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Documents
documents = [
    "Artificial Intelligence and Machine Learning",
    "Machine Learning and Artificial Intelligence",
    "Football World Cup"
]

# Create vectorizer
vectorizer = TfidfVectorizer()

# Convert into TF-IDF
X = vectorizer.fit_transform(documents)

# Calculate cosine similarity
similarity = cosine_similarity(X)

# Print similarity matrix
print(similarity)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample documents
documents = [
    "data science",
    "machine learning",
    "deep learning"
]

# Create vectorizer using unigram and bigram
vectorizer = TfidfVectorizer(ngram_range=(1,2))

# Generate TF-IDF matrix
X = vectorizer.fit_transform(documents)

# Print feature names
print(vectorizer.get_feature_names_out())

# Print matrix
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Training documents
train = [
    "Python programming",
    "Machine learning",
    "Data Science"
]

# Create vectorizer
vectorizer = TfidfVectorizer()

# Learn vocabulary from training data
vectorizer.fit(train)

# New document
test = [
    "Python Data"
]

# Convert test document into TF-IDF
test_vector = vectorizer.transform(test)

# Print TF-IDF values
print(test_vector.toarray())
```
