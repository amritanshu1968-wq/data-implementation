```python
import matplotlib.pyplot as plt
```

```python
categories = ["a", "b", "c", "d"]
values = [25,46,89,23]

#find maximum values
max_value = max(values)

#assign the colours
colours = ["black" if value != max_value else "red" for value in values]

plt.bar(categories,values,color = colours,width = 0.8,bottom=0)
plt.title("Bar chart:compare categories",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"Arial"
})

plt.xlabel("Categories",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"Arial"
})

plt.ylabel("Values",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"Arial"
})
```
