Step 1: Bar Cahrt - compare Categories

```python
import matplotlib.pyplot as plt
```

```python
categories = ["a", "b", "c", "d"]
values = [25,46,89,23]

#find maximum values
max_value = max(values)

#assign the colours
colours = ["black" if value != max_value else "red" for value in values]

plt.bar(categories,values,color = colours,width = 0.8,bottom=0)
plt.title("Bar chart:compare categories",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"Comic Sans MS"
})

plt.xlabel("Categories",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"Courier New"
})

plt.ylabel("Values",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"Calibri"
})
```

```python
Domain  = ["DSA", "DS & AI", "FULL STACK", "JAVASCRIPT"]
SGPA = [8.5, 9.0, 6.0, 7.0]

max_value = max(SGPA)

colours = ["Blue" if value !=max_value else "red" for value in SGPA]

plt.bar(Domain,SGPA,color = colours,width = 0.8,bottom=0)
plt.title("Bar chart:compare domains",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"arial"
})

plt.xlabel("Domains",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"courier new"
})

plt.ylabel("SGPA",fontdict = {
    "fontsize":16,
    "fontweight":"bold",
    "color":"red",
    "family":"courier new"
})
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# create a sequence of 30 consecutive numbers starting from 2026-01-01

days = pd.date_range(start='2026-01-01', periods=10)

#generate 30 random values from a normal distribution with mean 0 and standard deviation 1
# compute their cumulative sum to simulate a random walk
values = np.cumsum(np.random.normal(0,1,10)) +50

plt.plot(
    days,            # x-axis values (dates)
    values,          # y-axis values (random walk values)
    marker ="*",     # draw a star marker at each data point

    color = "black",  # set the line color to black

)
# add title to the graph 
plt.title("Line Chart: Trend Over Time ")

# add a label to the x-axis 
plt.xlabel("Date")

# add a label to the y-axis
plt.ylabel("value")

# set the x-ticks to show every 3rd day
plt.xticks(days[::3])


#display the graph
plt.show()
```

pie chart - show percentage

```python
Labels  = ["Lofer Shoes", "Tablet", "PC", "Mobile"]
sizes = [38, 45, 56, 78]

plt.pie(sizes, labels =Labels, autopct="%1.1f%%", startangle=90)
plt.title("Pie chart: Percentage Share")
plt.show()
```

Histogram- show data distribution

```python

data  = np.random.normal(loc=60, scale=10, size=1000)

plt.hist(data, bins=30, color="black", edgecolor="red")
plt.title("histogram: Data Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

Scatter Plot

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [35, 42, 15, 36, 47, 81, 78, 25]

plt.scatter(study_hours, marks, color="red")

plt.title("Study Hours VS Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()
plt.show()
```

Box Plot - Detect outliers and spread

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    "Group A": np.random.normal(70, 5, 100),
    "Group B": np.random.normal(60, 15, 100),
    "Group C": np.random.normal(55, 8, 100),
})

```

```python
sns.boxplot(data=df)
plt.title("Box plot: Spread and Outliers")
plt.show()
```
