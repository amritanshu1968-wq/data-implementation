```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#first check list of functionality of seaborn
print(dir(sns))
```

```python
ft = sns.load_dataset('flights')
```

```python
ft
```

```python
#Bar plot (Average passengers by month)
sns.barplot(data=ft, x="month", y="passengers", hue="month", legend=False)
plt.xticks(rotation=30)
plt.title("Passengers Distribution")
plt.show()
```

```python
#violin plot : Month wise passengers
sns.violinplot(data=ft, x="month", y="passengers")
plt.xticks(rotation=45)
plt.title("Month wise passengers")
plt.show()
```

```python
#scatter plot : Year wise passengers:
sns.scatterplot(data=ft, x="month", y="passengers")
plt.xticks(rotation=70)
plt.title("Passengers Distribution by month")
plt.show()
```

```python
#histogram : passengers
sns.histplot(ft["passengers"], bins=10)
plt.title("Passengers Distribution")
plt.show()
```

```python
#KDE(Kernel Density Estimate) plot : only passnegers
sns.kdeplot(ft["passengers"], fill=True)
plt.title("Density plot")
plt.show()
```

```python
#heatmap (Most popular plot for flights Dataset)
pivot = ft.pivot(index="month", columns="year", values="passengers")

sns.heatmap(pivot, cmap="YlGnBu")
plt.title("Flights Heatmap")
plt.show()
```

```python
#pair plot since the dstaset has only one numeric feature (passenger),
#the pair plot is simple

sns.pairplot(ft)
plt.show()
```

```python
#count plot : count the numbers of records for each month

sns.countplot(data=ft, x="month", hue="month", legend=False)

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,8))

#1 line plot
plt.subplot(2, 2, 1)
sns.lineplot(data=ft, x="year", y="passengers")
plt.show()
```

```python
#line plot with hue

sns.lineplot(data=ft, x="year", y="passengers", hue="month")
plt.title("monthly passengers trend")
plt.show()
```

```python
#Relational plot: "year", "passengers"

month_order=[
    "jan", "feb", "mar", "apr", "may", "jun"
    "jul", "aug", "sep", "oct", "nov", "dec"
]
sns.relplot(
    data=ft,
    x="year",
    y="passengers",
    kind="line",
    hue="month",
    #hue_order=month_order,
    legend='auto'
)
plt.show()
```
