```python
def add(a,b):
    return a + b
result = add(5, 3)
print(result)
```

```python
def subtract(a,b):
    return a - b
result = subtract(10, 4)
print(result)
```

```python
def multiply(a, b):
    return a * b
result = multiply(5, 3)
print(result)
```

```python
def divide(a, b):
    return a / b
result = divide(15, 3)
print(result)
```

```python
def quotient(a, b):
    return a % b
result = quotient(20, 8)
print(result)
```
