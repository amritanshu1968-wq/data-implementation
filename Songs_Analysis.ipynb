```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("songs.csv", encoding="latin-1")

print(df.head())
```

```python
print("Shape (Rows, Columns):", df.shape)
```

```python
df.info()
```

```python
print("Missing Values:\n")
print(df.isnull().sum())
```

```python
print(df.describe(include="all"))
```

```python
# Remove duplicate videos
df = df.drop_duplicates(subset="video_id")

# Remove rows having missing title
df = df.dropna(subset=["title"])

# Fill missing descriptions
df["description"] = df["description"].fillna("No Description")

# Convert numeric columns
cols = ["viewCount","likeCount","commentCount","favoriteCount"]

for col in cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df["publishedAt"] = pd.to_datetime(df["publishedAt"])

print(df["publishedAt"].head())
```

```python
df["Year"] = df["publishedAt"].dt.year
df["Month"] = df["publishedAt"].dt.month
df["Day"] = df["publishedAt"].dt.day
df["DayOfWeek"] = df["publishedAt"].dt.day_name()
df["Hour"] = df["publishedAt"].dt.hour

df[["publishedAt","Year","Month","DayOfWeek","Hour"]].head()
```

```python
total_videos = len(df)
total_views = df["viewCount"].sum()
total_likes = df["likeCount"].sum()
total_comments = df["commentCount"].sum()
avg_views = df["viewCount"].mean()

print("Total Videos :", total_videos)
print("Total Views :", total_views)
print("Total Likes :", total_likes)
print("Total Comments :", total_comments)
print("Average Views :", avg_views)
```

```python
top_views = df.sort_values("viewCount", ascending=False)

top_views[["title","channelTitle","viewCount"]].head(10)
```

```python
top_likes = df.sort_values("likeCount", ascending=False)

top_likes[["title","channelTitle","likeCount"]].head(10)
```

```python
channel_views = (
    df.groupby("channelTitle")["viewCount"]
      .sum()
      .sort_values(ascending=False)
)

print(channel_views.head(10))
```

```python
channel_videos = (
    df.groupby("channelTitle")["video_id"]
      .count()
      .sort_values(ascending=False)
)

print(channel_videos.head(10))
```

```python
yearly_uploads = df.groupby("Year")["video_id"].count()

print(yearly_uploads)
```

```python
df["YearMonth"] = df["publishedAt"].dt.to_period("M")

monthly_uploads = df.groupby("YearMonth")["video_id"].count()

print(monthly_uploads)
```

```python
top_comments = df.sort_values("commentCount", ascending=False)

top_comments[["title","commentCount"]].head(10)
```

```python
df["EngagementRate"] = (
    (df["likeCount"] + df["commentCount"])
    / df["viewCount"]
) * 100

df["EngagementRate"] = df["EngagementRate"].fillna(0)

top_engagement = df.sort_values(
    "EngagementRate",
    ascending=False
)

top_engagement[
    ["title","EngagementRate"]
].head(10)
```

```python
views = df["viewCount"].dropna().values

print("Mean :", np.mean(views))
print("Median :", np.median(views))
print("Maximum :", np.max(views))
print("Minimum :", np.min(views))
print("Standard Deviation :", np.std(views))
```

```python
df.to_csv("cleaned_songs.csv", index=False)

channel_views.to_csv("channel_views.csv")

top_views.to_csv("top_viewed_songs.csv")

top_likes.to_csv("top_liked_songs.csv")

print("Files Saved Successfully")
```
