```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
#movie review Dataset
reviews = [
    "I love this movie",
    "this movie is fantastic",
    "I hate this movie",
    "this movie is boring"
]
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
```

```python
X = vectorizer.fit_transform(reviews)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
tfidf_df = pd.DataFrame(
    X.toarray(),
      columns=vectorizer.get_feature_names_out()
      )

print(tfidf_df)
```

```python
new_review = ["I love fantastic movie"]

new_vector = vectorizer.transform(new_review)

new_df = pd.DataFrame(
    new_vector.toarray(),
    columns=vectorizer.get_feature_names_out()
    )
print(new_df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

reviews = [
    "Food was delicious",
    "Service was slow",
    "Delicious food and quick service",
    "Restaurant was clean"
]

vectorizer = TfidfVectorizer(stop_words='english')

X = vectorizer.fit_transform(reviews)

df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print(df)
```
