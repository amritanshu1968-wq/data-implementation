```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Python is used in data science",
    "Java is used for backend development",
    "Machine learning uses Python",
    "HTML CSS and JavaScript are frontend technologies"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

query = input("Enter your search query: ")

query_vector = vectorizer.transform([query])

scores = cosine_similarity(query_vector, X)[0]

ranked_documents = scores.argsort()[::-1]

for index in ranked_documents:
    print("\nScore:", scores[index])
    print(documents[index])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

reviews = [
    "This movie was excellent",
    "Amazing movie",
    "I loved this film",
    "Very good movie",
    "Worst movie ever",
    "Very boring film",
    "I hated this movie",
    "Terrible experience"
]

labels = [
    1, 1, 1, 1,
    0, 0, 0, 0
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(reviews)

model = LogisticRegression()

model.fit(X, labels)

new_review = ["This movie was amazing"]

new_vector = vectorizer.transform(new_review)

prediction = model.predict(new_vector)

if prediction[0] == 1:
    print("Positive Review")
else:
    print("Negative Review")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

documents = [
    "excellent movie",
    "amazing film",
    "very good movie",
    "I loved it",
    "terrible movie",
    "very bad film",
    "worst movie",
    "I hated it"
]

labels = [1, 1, 1, 1, 0, 0, 0, 0]

X_train, X_test, y_train, y_test = train_test_split(
    documents,
    labels,
    test_size=0.25,
    random_state=42
)

vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)

X_test_tfidf = vectorizer.transform(X_test)

model = LogisticRegression()

model.fit(X_train_tfidf, y_train)

predictions = model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Python is widely used for data science",
    "Machine learning is a branch of artificial intelligence",
    "Python is popular for machine learning",
    "Java is widely used for backend development",
    "React is used for frontend development",
    "Data science uses Python pandas numpy and machine learning"
]

vectorizer = TfidfVectorizer(stop_words='english')

X = vectorizer.fit_transform(documents)

query = input("Search: ")

query_vector = vectorizer.transform([query])

similarity = cosine_similarity(query_vector, X)[0]

ranking = similarity.argsort()[::-1]

print("\nSearch Results:\n")

for i in ranking[:3]:

    print("Similarity:", round(similarity[i], 3))
    print("Document:", documents[i])
    print()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

texts = [
    "excellent movie",
    "amazing film",
    "great movie",
    "good acting",
    "terrible movie",
    "bad film",
    "worst acting",
    "boring movie"
]

labels = [1, 1, 1, 1, 0, 0, 0, 0]

X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.25,
    random_state=42
)

vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

model = LogisticRegression()

model.fit(X_train_tfidf, y_train)

predictions = model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, predictions)

print("Predictions:", predictions)
print("Actual:", y_test)
print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(y_test, predictions)
recall = recall_score(y_test, predictions)
f1 = f1_score(y_test, predictions)

print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```
