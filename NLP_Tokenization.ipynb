```python
# Stop word removal
#is
#am
#are
#an
#of
#to
#in
#for
#and

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "I am leaning Natural Language Processing"

stop_words = set(stopwords.words('english'))

tokens = word_tokenize(text)

filtered_tokens = [token for token in tokens if token.lower() not in stop_words]

print(filtered_tokens)
```

#Stemming

Stemming reduces words to their root form by removing prefixes or suffixes using simple rules. The resulting word (called a stem) may not be a valid dictionary word.

# | Original Word | Stem    |
# | ------------- | ------- |
# | Playing       | Play    |
# | Played        | Play    |
# | Plays         | Play    |
# | Studies       | Studi   |
# | Running       | Run     |
# | Connected     | Connect |

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["mornings", "played", "studies", "running"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

#Lemmatization

Lemmatization converts words into their dictionary (base) form, called the lemma, using vocabulary and grammar.

# | Original Word | Lemma |
# | ------------- | ----- |
# | Running       | Run   |
# | Studies       | Study |
# | Better        | Good  |
# | Mice          | Mouse |
# | Children      | Child |

```python
import nltk

nltk.download()
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "studies", "better", "mice"]

for word in words:
   print(word, "→", lemmatizer.lemmatize(word))
```
