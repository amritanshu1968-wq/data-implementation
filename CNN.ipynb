```python
from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# 1. Sequential: Iska matlab hum layers ko ek ke baad ek line se jodh rahe hain
model = Sequential()

# 2. Conv2D: Ye tumhara pehla pillar hai (Feature Extractor)
# Yahan 32 filters hain, jinka size 3x3 hai. Activation 'relu' lagaya hai.
model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 3)))

# 3. MaxPooling2D: Ye tumhara dusra pillar hai (Size Reducer)
# Ye 2x2 ke box mein se sabse badi value chunega aur image ka size aadha kar dega.
model.add(MaxPooling2D(pool_size=(2, 2)))

# --- Tum chaho toh aisi hi aur layers bhi jodh sakte ho ---

# 4. Flatten: Ye matrices ko 1D vector (ek lambi seedhi line) mein badal deta hai
model.add(Flatten())

# 5. Dense: Ye tumhara Fully Connected Layer (Classifier) hai
# Pehle hidden layer mein 128 neurons hain
model.add(Dense(128, activation='relu'))

# Final Output Layer: Agar Cat vs Dog classify karna hai (2 classes)
# Softmax ya Sigmoid function probability batayega
model.add(Dense(1, activation='sigmoid'))  # Binary classification ke liye sigmoid

model.compile(
    optimizer='adam',
      loss='binary_crossentropy',
        metrics=['accuracy']
    )

model.summary()  # Ye model ka structure aur parameters dikhayega
```
